- Definition for each variable:

  - candidate_key、candidate_name、county_city、electoral_district：研究名冊的候選人鍵、姓名、縣市與選區，依姓名＋完整選區唯一對應索引。
  - 教育線索的upstream_claim_raw／source_type／source_year：第三方原文字串與分類，明確不是已驗證教育欄；source_url／fetch_url分別保留含頁碼原連結及實際請求網址。
  - http_status、source_file、source_sha256、parse_error：存取、保存與解析結果；document_is_government依實際host，未照抄第三方gov標記。
  - requested_page、actual_name_pages、page_name_status、html_name_present、validation_status：指定頁、實際姓名出現頁與檢查結果；只驗證姓名可讀，不當作所有欄位已正確切分或當屆身分認證。
  - 頁面表page／page_text：完整來源頁與轉出的整頁文字（換行以字面\n保存），page_role明示未切分候選人欄位；可能含同頁其他候選人資料，不能整頁歸屬一人。
  - 新人物頁profile_url／education_raw／experience_raw：已查看的明確欄位或逐字原文；source_year／evidence_tier／identity_status保留年代、自述與待獨立身分核對。
  - 新逐欄證據field／value_raw／source_locator／source_sha256：欄位、原文、DOM區塊或手選文字片段及來源檔hash；不擅自補學位或畢業。林鈺梅birthday為官網明載自述，不混入官方歷史支持欄。
  - 社群account_url／status：人物頁專屬社群區塊連結，目標內容未核對；縣市長第三方線索kind／url／upstream_evidence_url／note／verification_status保留第三方標記，不宣稱本研究認證。
  - 社群current工作單每人每平台一列；official_urls為來源頁連出的入口，account_status保留目標內容待核對。window_count與followers_start/end均未蒐集、留空白。未解析分享及群組網址另表保留，不視為個人帳號。
  - upstream_repository／upstream_commit：上游公開索引固定版本，供來源追溯及署名。
  - current主表／欄位表沿用舊變數定義，本輪更新新增人物頁、教育來源待核對狀態、1筆birth_date與社群來源連結；其餘未改欄位經assertions比對。議員背景新增additional_profile_url／status及birth_date_self_declared／source，歷史支持690人欄位保留不動。
