- Definition for each variable:
  - candidate_key／candidate_name／county_city：原2026名冊鍵、姓名及縣市；同名不同縣市不能直接合併。
  - cells.name_cell／birth_roc_raw／sex_raw／education_raw／experience_raw：指定公報表格欄位逐字抽出原文；換行以字面\n保存，一筆一個實體CSV列。
  - cells.row_top／row_bottom／x_shift：已視讀表格列的精確PDF座標；出生、性別、學歷、經歷各自限定欄位，結束於政見欄以前。座標與期望出生／性別核對值固定在處理程式CELL_REVIEW，不以跨欄整頁文字猜測。
  - expected_birth／expected_sex：AI查看原表格影像後的核對值，仍待學生檢閱；不是新增獨立資料來源。
  - birth_date_declared／birth_year_declared：官方公報民國年加1911，經日期合法性檢查；不是全部已連到2026個人的完成值。
  - source_url／source_file／source_sha256／page／source_year：來源URL、保存原檔、固定hash、1-based頁碼及年代。
  - identity_status／history_same_origin_note／same_district_number：身分連結支持程度、同源限制與跨選區旗標。另行核對之履歷、媒體與公報以完整來源鏈保留。
  - profile.role_2026／area_2026／election_history_raw：人物頁明確當屆角色、選區與歷史參選區塊；不能用過去選區推定2026任職。
  - profile.accepted_for_source_inventory：是否可作背景來源線索，不是學位或戶籍認證。False包括占位、同名地區不同、正文不全及缺少當屆身分情境。
  - evidence.field／value_raw／source_locator：逐欄值及實際DOM或PDF座標；identity_context為來源區塊原文片段／欄位串接，用來回查身分，不直接當分析特徵。
  - council新增birth_date_bulletin_candidate／birth_year_bulletin_candidate／sex_bulletin_candidate／education_bulletin_raw／experience_bulletin_raw：13人的歷史候選值；bulletin_source_url／file／page與bulletin_identity_status跟值一起保存。這些欄位不能直接等同birth_date_supported_history等已支持欄。
  - supported_history的林于凱：2026人物頁同姓名、同高雄第7選區，兩個中山大學系所與歷史公職履歷和公報一致。黃敬雅：多所學校順序及復興航空工作一致，另以2026採訪確認同一職涯者投入同選區，保留政黨轉換年代。
  - social新增source_year／source_identity_status：新人物頁的年代與情境；舊證據沒有新增來源年份時留空，不推定2026。只取ul.candidate-social-links；組織頁尾、分享、群組、貼文不作個人帳號。
