- Definition for each variable:

  - 共通 candidate_key、candidate_name、county_city、office_type：研究名冊識別與人名、公職、縣市。
  - 證據表 field、value_raw：人物頁入口／學歷／一般經歷／明列曾參選的原文或0值；profile_url 入口與背景欄位分列。
  - source_url、source_file、source_sha256、source_locator：原文出處、保存檔與定位。evidence_tier：競選／政黨申報，未獨立認證。identity_status：唯一姓名＋縣市＋公職對應、仍待獨立身分核對。observation_date：本次整理日。
  - 人物頁表 profile_url、education_raw、experience_raw：可點選頁與未分級的學經歷原文；birth_date、sex 留空；對應 status 表示背景區塊未列示。previous_moi_profile_present 表示是否已在原692筆人物頁表，不等於現任或已認證。
  - 主表 registration_* 為原登記出處；party_registration_raw 原推薦政黨；qualification_* 保留既有資格狀態、未在本輪升級；sample_inclusion 保留組內規則待定；directory_lead_count 保留舊名錄計數。各 *_status 為目前線索狀態，不能當作分析值。
  - 欄位清單 field、status、existing_value、evidence、note、next_source、updated_date：欄位、查核狀態、來源值、出處、限制、下一步與日期；只更新精確對應的profile／education及有明文依據的first_time欄位。
