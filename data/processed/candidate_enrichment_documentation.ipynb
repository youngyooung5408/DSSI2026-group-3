### Definition for each variable

遠見媒體彙整檔逐一核對81人姓名、縣市、登記推薦政黨（明列縮寫與温／溫異體字），吻合不代表逐項學位已查證。空陣列與「尚未公布」仍為缺值。遠見原文學歷保存於observations之education_gvm_compilation；僅在原CNA學歷空白時補入education_media_supplement並明標學位待核實。新增羅佩秦學歷說明，原視報導可支持陸官正期53期畢業證書的說法，但未獨立確認其兩項學士學位；不改education_2026或教育分類。出生年無新增，仍缺羅佩秦、魏造文。GVM政見為媒體編輯摘要，不合併到候選人官網原文段落。

原名冊欄位沿用 [第一版字典](candidates_2026_documentation.ipynb)。新增欄位：
- birth_date_verified / birth_date_source_url / birth_date_evidence_status：ISO 出生日、官方來源及核對層級；CEC 與媒體完整生日一致者也明列媒體輔助核對，不假稱本次公報認證。
- sex_2026 / sex_source_url：公開來源性別標記；本次再查原缺性別的25人，新增19筆可追溯觀察（政府／議會／政黨／專業學會8人、媒體11人），主表80/81（新增5人依使用者提供，未獨立查證）；其中葉人文依媒體「大港男兒」明示描述，並非戶籍性別欄。原缺6人中，使用者補充王肇民、洪方隆、朱定瑀、羅佩秦、李吳穎智為男，以user_provided標記；魏造文回覆未知，主欄仍保留空白，sex_evidence_status記錄user_reported_unknown。社群目錄暫存值保留原來源。李文良僅取得官方名冊搜尋索引片段，林國漳採web工具抽出PDF文字（直接下載403），限制分列於sex_evidence_status。性別欄為公開來源標記，不代表已查驗2026戶籍資料。逐人來源與未解缺口見data/raw/sex_verification/sex_search_audit_25.csv。
- sex_source_tier / sex_evidence_status：逐人來源層級、身分核對及來源限制；sex_directory_supplement / sex_directory_source_url：未獨立核實的社群目錄值，不能當作主欄已完成。
- age_2026_minus_birth_year：2026 減已核對出生年；不是投票日足歲。media_birth_year 只是媒體輔助觀察，未拿來自動填 verified。
- education_profile_as_captured / education_profile_source_url：官署、政黨或正式個人頁於擷取時刊載的學歷文字，底層日期可能較早，保留博士生／肄業等詞。education_media_supplement 是媒體補充；education_latest_cec / year 是配對後最近 CEC 歷史學歷／年份。education_2026 暫不混填三者，仍待當屆公報核實。
- media_birth_original / media_birth_year / media_experience_supplement：來源原文、可識別年份及經歷補充，不能當成官方已確認現在職務。birth_date_media_supplement 僅轉換格式完整有效的媒體生日；age_2026_from_media_birth_year = 2026 減媒體出生年，獨立保留補充來源標籤，不冒充 primary-derived 年齡。
- incumbency_2026：目前政府首長網站可支持的現任觀察；未確認的挑戰者不自行填值。
- official_profile_url / official_campaign_website / official_facebook_url / official_facebook_link_evidence：官署人物頁、候選人自設官網、候選人或政黨頁直接連出的 Facebook 及該連結依據。Municipal footer 的機關粉專已排除。
- qualification_status / qualification_source_url：10/16 官方審查通過片段及原頁 URL；sample_inclusion 仍為組內待定。
- first_time_any_elected_office / first_time_mayor_lifetime：正面歷史證據可否定首次時才填 0；沒有資料不填 1。prior_any_office_evidence_url / prior_mayor_evidence_file 為支持來源；兩個 first_time_*_evidence_status 記依據。first_time_review_status 顯示仍待補足的終生查證。
- no_prior_mayor_match_in_available_archive：1=目前配對成功的可得紀錄中未見縣市長參選，0=有。**不是終生首次**，未核實身分者也可能沒有成功配對。
- history_identity_verified / identity_conflict_notes：跨來源生日核對層級與排除同名紀錄數；細節在 history 的 identity_match_status。
- background_collection_status：已有部分來源或搜尋後仍無可信背景，不宣稱完成全部候選人資料。

觀察表欄位 candidate_key/name、field/value 為人物及來源欄位值；source_type/url/file/locator/time_context、captured_date、verification_status 為來源類型、定位與適用時間。history 的 election_year/context、office_is_county_city_mayor、sex/birth_date/birth_raw/education_level_cec、geography_codes/elected_marker 為原始紀錄解析，source_file/row/archive_member/url 定位原檔，identity_match_status 保留採納與排除理由；elected_marker 不拿來自動推定 2026 現任。

campaign_sources 的 source_kind/url 為來源類型與連結，discovered_on/source_file 是連結證據，verification_status 與 content_capture_status 區分已下載的官網和只有連結的社群。text_blocks 的 block_number/html_tag/text_original 為擷取段落定位與原文，published_date 未能核實留空，coding_status 表未編碼；這些不是已分割的獨立政見，不能直接計算 S5–S7。文字可能含 UI 或捐款條款，須依編碼手冊後續篩選。

coverage 的 *_available / *_found / *_resolved 是有無資料（0/1），不是該人的政治屬性；campaign_text_blocks 是已存段落數；facebook_posts_collected=0 只表示本次未蒐集貼文，window_post_count_status 明示不能解讀為候選人零貼文。remaining 列仍缺欄位。search_log 的 query/source_file/status 記真實查詢；issues 的 field/issue/source_file 記日期等來源品質問題。

```python
from pathlib import Path
import hashlib
import pandas as pd


def find_project_root(start=None):
    start = Path.cwd() if start is None else Path(start)
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate / "AGENTS.md").is_file() and (candidate / "templates").is_dir():
            return candidate
    raise FileNotFoundError("Open this notebook within the student project containing AGENTS.md and templates/.")


PROJECT_ROOT = find_project_root()


def project_file(relative_path, area, must_exist=True):
    # Explicit project-relative paths only; never select the first matching file.
    rel = Path(relative_path)
    if rel.is_absolute() or ".." in rel.parts or "<" in str(rel):
        raise ValueError(f"Replace the placeholder with an exact project-relative path: {relative_path}")
    path = (PROJECT_ROOT / rel).resolve()
    allowed = (PROJECT_ROOT / "data" / area).resolve()
    if allowed == PROJECT_ROOT or not allowed.is_relative_to(PROJECT_ROOT):
        raise ValueError(f"data/{area}/ resolves outside the project root.")
    if not path.is_relative_to(allowed):
        raise ValueError(f"Expected a file inside data/{area}/: {relative_path}")
    if must_exist and not path.is_file():
        raise FileNotFoundError(path)
    return path


def sha256_file(path):
    digest = hashlib.sha256()
    with path.open("rb") as stream:
        for block in iter(lambda: stream.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()

def verify_input_sha256(path, expected):
    # Bind this run to the exact file inspected when the notebook was prepared.
    if (not isinstance(expected, str) or len(expected) != 64
            or any(char not in "0123456789abcdefABCDEF" for char in expected)):
        raise ValueError("Fill the expected SHA-256 from the inspected input before loading data.")
    actual = sha256_file(path)
    if actual != expected.lower():
        raise ValueError(f"Input changed: identify the new version before loading {path.name}.")
    return actual
```
