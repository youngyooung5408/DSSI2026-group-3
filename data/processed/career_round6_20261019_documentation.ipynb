- Background: 第六輪候選人資料整合：18人背景來源原文（其中16個新人物頁）、4人新增官方生日支持、3人新增歷史申報性別、5人補非首次任何民選公職，並整合29組社群入口與81人首長比對；另續補陳重嘉與温世政完整生日（後者專業學會來源非政府）；1162列第三方履歷主張保留為待查。
- Unit of observation: master／council／mayor／profile為候選人；fields為候選人×欄位；社群工作單為候選人×平台；帳號證據為候選人×平台×來源×URL；背景證據為候選人×欄位×來源；worklist為上游履歷主張×來源。
- Sample period: 2026候選人登記名冊為框架。背景引用2014、2018、2020、2022、2024歷史申報及2026擷取的人物頁，不當作同時期的戶籍或學位認證；貼文窗口未啟用。
- Known issues: 仍733人缺目前幾類人物頁；歷史姓名匹配需身分證據。692人的2022歷史生日／性別序列不混入本輪跨年資料；4人額外官方生日與3人性別分欄。林德建未取得明載性別，維持未知。政黨／競選頁的履歷為自述、學位未認證。524人的1162列主張全未核實，不以姓名出現在整頁文字當作完成。社群目標內容未全面核對，貼文、窗口量及追蹤數缺漏不是0。首長名錄不是停職／代理法律狀態全查；宜蘭另記林茂盛代理，雲林僅名錄來源、近期直接政府內文待補。非名錄首長不能直接推定任期屆滿或挑戰者。
- Definition for each variable:
    - `candidate_key`沿用名冊公職×縣市×選區×姓名／既有縣市長ID，不以姓名單欄連接；`office_type`與`county_city`保留名冊原文。
    - `birth_date_supported_history`／`sex_supported_history`為原2022有身分支持的歷史序列；`birth_date_additional_official`、`birth_date_additional_source_year`為額外官方人物頁或跨年中選會生日；`sex_additional_source_value`／`sex_additional_source_year`為明載申報性別，與性別推測分開。`birth_date_self_declared`為先前自述，不混入官方。
    - `additional_historical_education_cec`為新連結歷史表原學歷欄，不認證2026最高學歷；`education_raw`／`experience_raw`保留逐頁原文；`round6_profile_urls`為本輪已讀來源，原profile欄不覆寫。
    - `first_time_any_office_supported`／`first_time_this_office_supported`：0表示有可支持過往參選／任職的來源；未知維持空白，未由查無紀錄填1。
    - `matches_listed_elected_head`僅為同縣市候選人姓名與名錄民選首長的比對0/1；`government_role_name_as_captured`保留另一官方頁明載名稱；`acting_head_observed`僅宜蘭=1，其餘空白不等於沒有代理；`listed_elected_head_registered_in_roster`只表示登記名冊有無同名民選首長，不代表最終資格。
    - `account_url`為候選人專屬區塊直接列出的網址；`source_locator`保留DOM／明載文字定位；`target_content_status`目標帳號內容存取狀態，不等同來源頁成功。`window_count`與追蹤起訖維持空白。
    - 待查表`upstream_title`／`upstream_organization`／`upstream_*date`為第三方主張，不是本研究核實職務；`http_status`與`page_mentions_candidate_name`只記來源可得性；`claim_verified=0`表示尚未核實主張。
    - 各輸出其他欄位沿用前輪已記錄定義；來源、年份與證據狀態保留在逐欄證據表，未將概略履歷轉成行政年資或政治家族判定。
