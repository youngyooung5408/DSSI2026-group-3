### 完整欄位定義

| Variable | Definition |
| --- | --- |
| `ARI_mean` | 100次重排列順序的平均ARI。 |
| `ARI_min` | 相對固定鍵排序基準的最小adjusted Rand index，1代表相同分割。 |
| `B` | bootstrap重抽次數500。 |
| `G5` | profile_id中G5的0／1類別。 |
| `G5_audit_status` | 上游G5查核狀態，無重新提升證據等級。 |
| `G5_first_time_any_elected_office` | 0曾參選任何民選公職；1按楊修訂規則首次，未當選也算曾參選。 |
| `G5_label` | 上游G5文字標籤，原樣保留；分析以明定0／1為準。 |
| `G5_source` | 上游G5來源網址或引用資訊。 |
| `G6` | profile_id中G6的0／1類別。 |
| `G6_evidence_tier` | 上游G6證據層級，含次級來源，原樣保留。 |
| `G6_label` | 上游G6文字標籤，原樣保留；0仍須解釋為搜尋未發現。 |
| `G6_political_family` | 1有符合定義的公職親屬；0完成上游紀錄搜尋未發現，非不存在的證明。 |
| `G6_review_note_tk_20261108` | 上游關係推論或一般軍文職範圍排除等註記。 |
| `G6_source` | 上游G6來源網址或搜尋紀錄。 |
| `absent_fraction` | absent_n/B；小群重抽缺席率，非正式推論或編碼可靠度。 |
| `absent_n` | 該原群整群沒被抽到的重抽次數。 |
| `actual_k` | 替代切樹實際非空群數，需等於requested_k。 |
| `all_clusters_n_ge_8` | 是否每群至少8人，對照提案約8–10人下界，非硬性顯著水準。 |
| `candidate_key` | 上游候選人唯一鍵；合併／首長／議員池可連結，不跨本屆另造人。 |
| `candidate_name` | 原候選人姓名，識別用，不入距離。 |
| `cluster_id` | 替代K的本地C1、C2…；不可直接跨K或池視為相同類型。 |
| `cluster_profile_id` | 主群固定組合碼P00／P01／P10／P11，兩數字依序是G5與G6。 |
| `cluster_profile_label` | 主群可讀標籤；G6=0稱搜尋未發現公職親屬。 |
| `cluster_profiles` | 群內組合以+連接；K診斷各群以|分隔。 |
| `cluster_sizes` | 各本地C群人數，以|分隔，順序與cluster_profiles一致。 |
| `clustering_pool` | 逐人主結果所屬池：all／mayor／council；三池重複列不能當成418人。 |
| `county_city` | 臺北市或高雄市，僅核對樣本，不入距離。 |
| `electoral_district` | 上游選區，識別與後續連結用，不入距離。 |
| `is_primary` | 該K是否等同零距離主群。 |
| `jaccard_common_support_mean_when_present` | 只限當次抽中原群成員、且非空時的最大Jaccard平均；本題按同一組合必為1。 |
| `jaccard_full_universe_mean` | 原群全體成員與重抽各群最大Jaccard之500次平均；受未抽中成員影響。 |
| `k` | 替代K或列順序診斷的實際要求群數。 |
| `max_cluster_n` | 替代切樹最大群人數。 |
| `min_cluster_n` | 替代切樹最小群人數。 |
| `missing_fields` | 被缺值篩選排除的G5／G6精確欄名，以;分隔；此檔0列。 |
| `n` | 該池該組合人數；保留0人格，不計入實際群數。 |
| `observed` | True表示n>0，False為未出現組合。 |
| `office_type` | 直轄市長或直轄市議員，三池分割依據，不入距離。 |
| `original_n` | 重抽前原群人數。 |
| `percent` | 100*n/pool_n，為池內百分比，非估計母體參數。 |
| `permutations` | 列順序重排次數100，不是顯著性排列檢定。 |
| `pool` | 資料所屬池all／mayor／council，不是地理分群。 |
| `pool_n` | 該池的保留候選人總數，209／11／198。 |
| `primary_k` | 該池距離0切群之實際非空群數：合併4、首長3、議員4。 |
| `primary_profile_cut` | 該K是否為本池主群數。 |
| `primary_silhouette` | 該人在主分群之silhouette，本次皆1；同組合重複造成，非外部驗證。 |
| `profile_id` | 固定組合碼P00／P01／P10／P11。 |
| `profile_label` | 固定組合可讀標籤。 |
| `reason` | 排除原因missing_G5_or_G6；本次沒有任何排除。 |
| `requested_k` | 替代切樹要求的K，從2到該池非空組合數。 |
| `same_partition_fraction` | 100次重排中ARI近似1的比例，非信賴度。 |
| `seed` | 固定亂數種子20261108，重抽與列順序各自初始化。 |
| `silhouette_mean` | 使用預先計算Hamming距離矩陣之平均silhouette。 |

JSON格式：output_manifest為逐CSV的path／rows／columns／sha256／verification清單；validation記錄分析特徵、精確輸入pins、篩選人數、主群數、實際通過檢查、套件版本、固定seed與已知限制。
