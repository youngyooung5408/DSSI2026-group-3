- Definition for each variable:

- candidate_key、candidate_name、county_city、office_type：研究名冊識別與姓名、縣市、公職。
- platform：Facebook／Instagram／Threads；account_url／official_urls：個人入口，後者以 | 分隔多個保留網址，並非全部獨立帳號。
- source_url、source_file、source_locator／discovered_on：發現入口的 URL、原始檔、候選人欄位或區塊定位。
- evidence_status：party_candidate_object_link（政黨候選人物件）、candidate_website_link（本人競選網站）、line_official_account_self_link（LINE 官方帳號自列）、campaign_directory_candidate_link（競選聯盟人物頁，仍待獨立核對），或 inherited_ 前綴（舊表原狀態）。
- target_content_status：未核對目標頁內容；checked_date／updated_date：本次來源整理日期。
- account_status：有來源連結、內容待核對，或尚未找到而非不存在。
- posts_status：尚未蒐集而非零；window_count、followers_start、followers_end：留空的後續欄位。
- unmatched 表 name、city、source、reason：來源人物、縣市、網址及沒有唯一研究名冊對應之理由，不代表人物不存在或不參選。
