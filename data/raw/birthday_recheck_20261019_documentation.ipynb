# Data Description
- Background: 中選會保存之彰化縣選委會2014年彰化市長選舉公報，補邱建富生日來源；另保存本輪逐格人工查核紀錄。
- date received: 20261019
- Path to data file: https://web.cec.gov.tw/api/file/1f07fea7-6589-448f-b00d-4d17624878b2.pdf
- Unit of observation: PDF每一候選人欄；JSON為逐格查核紀錄，本輪1人。
- Sample period: 2014年選舉資料；2026-10-19擷取與查看。
- Known issues: 公報個人資料為候選人填報；不是戶籍或學位認證。本輪只核對第1頁號次1，不宣稱其他所有表格已逐格核對。
- Definition for each variable:
  - candidate_key、candidate_name：研究候選人鍵與姓名。
  - source_url、source_file、source_sha256、source_year、page、source_locator：原始來源、hash、年代、頁碼與表格定位。
  - birth_date_raw、birth_date、sex_raw：生日原文、轉西元完整日期、表格性別原文。
  - verification_method、historical_match_files、checked_date、limitation：查核方法、歷史對照檔與列號、日期與限制。
- Data path: data/raw/birthday_recheck_20261019/qiu_2014_bulletin.pdf；data/raw/birthday_recheck_20261019/visual_review.json。
- Role: raw source / manual source observation。
- Verification status: PDF已本地render並查看姓名及生日欄，以下核對保存bytes、頁數與查核紀錄。
- SHA-256: 各檔固定值見程式。
- Next processing notebook: data/code/verify_mayor_birthdate_20261019.ipynb


```python
from pathlib import Path
import pandas as pd,json,hashlib
ROOT=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'AGENTS.md').exists() and (p/'templates').is_dir())
def sha(p):return hashlib.sha256(p.read_bytes()).hexdigest()

import pypdfium2 as pdfium
```
