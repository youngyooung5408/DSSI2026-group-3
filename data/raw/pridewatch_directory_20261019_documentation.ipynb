# Data Description
- Background: 從PrideWatch公開問卷及公開候選人目錄取得人物頁出處，依2026名冊姓名找到31份公開人物頁，再核對實際姓名／選區／年代。另保存2份2026媒體報導以連接歷史與當屆參選情境。
- date received: 20261019
- Path to data file: data/raw/pridewatch_directory_20261019/manifest.json；原始URL、保存檔與原bytes SHA逐筆列示。
- Unit of observation: 每個公開URL的回應檔與存取紀錄。39個URL含31個人物頁、2個媒體頁、問卷與目錄導覽回應。
- Sample period: 目錄標示2022、人物頁部分有2026參選欄；內容年代各異。本輪2026-10-19保存，不能以取得日期認定當屆任職或最高學位。
- Known issues: 目錄的姓名GET查詢仍回傳多位候選人，未把它當作伺服器已篩選或身分已驗證；本研究只把可見連結當線索，逐頁再檢查。PrideWatch為公民團體頁，原文可為候選人申報或歷史內容，不當作政府認證。李文有占位文字；竹北陳冠宇不可接到高雄同名者；黃瓊慧回應未有可用人物名稱區塊。HTTP200與姓名命中不足以認定背景有效。FTV一般Python TLS驗證失敗後以正常curl完成下載，未停用TLS驗證。
- Definition for each variable:
  - url／status／path／sha256／captured_date：實際請求、HTTP狀態、保存檔、原bytes hash與取得日。
  - candidate_name（若有）：請求用途／姓名線索，不代表回應已認證身分；兩份媒體的文字含用途註記。
  - return_code（部分curl紀錄）：curl實際結束碼。
  - retrieval_plan.profiles：目錄可見姓名→可見人物頁URL的派生取得計畫，不冒稱原始候選人欄位。
- Data path: data/raw/pridewatch_directory_20261019/；既有官方PDF沿用background_expansion_20261019保存檔，逐人路徑／hash見公報欄位表。
- Role: raw public source captures / acquisition metadata；未更改原bytes。
- SHA-256: manifest及plan固定值見PINS；各來源固定值見manifest。
- Verification status: 39個回應保存檔與hash、HTTP200、URL唯一；內容是否可用另在processed逐頁判斷，未以200當完成。

```python
from pathlib import Path
import json,hashlib,pandas as pd
ROOT=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'AGENTS.md').exists() and (p/'templates').is_dir())
def sha(p):return hashlib.sha256(p.read_bytes()).hexdigest()
def read(p):return pd.read_csv(ROOT/p,dtype=str).fillna('')
```
