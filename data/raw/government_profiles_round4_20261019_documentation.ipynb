```python
from pathlib import Path
import json,hashlib,pandas as pd
ROOT=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'AGENTS.md').exists() and (p/'templates').is_dir())
def sha(p):return hashlib.sha256(p.read_bytes()).hexdigest()
def read(p):return pd.read_csv(ROOT/p,dtype=str).fillna('')
```

## Change for Your Own Data

```python
assert sha(ROOT/'data/raw/government_profiles_round4_20261019/manifest.json')=='cd41b4cb7473fcc0b65497b85e8f460c9ed31b49990f298246b27ae308377a72'
records=json.loads((ROOT/'data/raw/government_profiles_round4_20261019/manifest.json').read_text())
assert len(records)==3
for r in records:
 assert r['status']=='200' and sha(ROOT/r['path'])==r['sha256']
 if r['content_type']=='application/pdf':assert (ROOT/r['path']).read_bytes().startswith(b'%PDF')
df=pd.DataFrame(records)
print('PASS：3個來源保存檔與hash；日期層級分開。')
```
