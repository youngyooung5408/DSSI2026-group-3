# Data Description
- Background: 魏造文專項調查，回應使用者附圖與男的編碼決定，另查明載性別的文字欄位、學經歷、歷史選舉、公開政見與親屬線索。
- date received: 20261108（本輪實際下載及調查）
- Path to data file: data/raw/wei_zaowen_20261108/，原始網址與SHA見manifest及下方逐檔PINS。
- Unit of observation: 一份原頁HTTP回應、API回應、文字擷取，或人工查核紀錄。
- Sample period: 2026地方選舉；保存歷屆檔案另按原年份。
- Known issues:
    - 選舉黃頁為第三方可編輯資料；頁面男與API m是同一資料源，不當兩份獨立證據，也非戶籍認證。
    - 使用者附圖決定另記於review.json，本地未另有截圖檔，不製造不存在的圖檔連結；助理未由外觀推斷性別。
    - 年齡／生日未查到；教育未公開仍按使用者規則歸最低類並留旗標；不能描述為已證明低學歷。
    - G6=0是指定搜尋未發現，不是證明無政治親屬。145份歷屆CSV完全同名未命中，不能排除更名及資料未涵蓋的選舉。
    - LTN只讀魏造文本人區塊，其餘候選人資料未用於本案；新聞全文與來源已保存。未補社群帳號或貼文。
- Definition for each variable:
    - manifest：source_id、url、http_status、path、sha256、retrieved_date、full_text_reviewed、review_scope。
    - API Candidate：name／gender／no／birth／education／experience／is_present，僅指該網站記錄。
    - review.json：G1–G6逐欄判定、查詢詞、排除線索、原始歷屆檔指紋及限制。
- Data path: 下方PINS的每一精確檔案。
- Role: raw / source capture / documented manual review
- Verification status: clean kernel逐檔SHA、JSON格式與姓名、公職地域、號次、gender欄位核對；原始bytes未改。
- SHA-256: 下方逐檔輸出。
