## What does this file do?

- 依 2026-11-08 組內決定的分析範圍，建立一份 G1–G6 合併檔：全部直轄市長與縣（市）長（81 人），加上臺北市議員（98 人）與高雄市議員（101 人），共 280 人；其他縣市議員不納入。觀察單位為「候選人」，鍵為 `candidate_key`。
- 以 `candidates_g1_g6_current.csv` 為底，只在原值為空時，套用楊喬茵 2026-11-04 審查檔中 `decision` 為 `accepted` 或 `accepted_group_decision` 的建議值（縣市長 G1/G2/G3/G5；臺北、高雄議員 G1/G3/G5），以及 `manual_check_worklist_taipei_kaohsiung_20261104.csv` 中已填 `found_value` 的人工查核值。審查檔與 current 沒有重疊的已填值，因此不會覆蓋任何既有資料。
- G5 補值同時更新 `G5_label`；每個 G1/G2/G3/G5 欄位新增 `*_fill_status`、`*_fill_source`、`*_fill_note`，記錄值來自 current、審查補值，或仍缺漏。何恕懷 G3 為學生登入 Facebook 抄錄，許以琳 2026-11-09 確認粉專與登記者為同一人，狀態標 `filled_student_check`。
- 另輸出 G6 人工查核清單：分析樣本中 G6 為「未知」者，附空白回填欄，供學生手動查核政治家族。
- G1 補值有部分是由報導「現年 XX 歲」推算，可能與「2026 − 出生年」差 ±1，依據寫在 `G1_fill_note`。
- G6 為「有記載／未知」，未知不是 0；本檔不補 G6，只保留原值。
- 再套用許以琳 2026-11-08／11-09 的決定檔：補入人工查得的 G1／G3／G5／G6（`*_fill_status = filled_student_decision`），並排除資料無法取得的候選人；`action = note` 的候選人層級備註（原住民、政治家族、爭議事件）寫入 `candidate_note`。李政憲、劉仙娥、余聖明的 G5 另以中選會歷屆候選人檔核對生日。排除者不進分析檔，另存 excluded 檔並保留排除理由（例如魏造文年齡不公開、林揚庭與李傅鈺婷為政治家族）。
- 新增 `G1_age_basis`：`birth_date`（由出生日期計算）、`reported_age`（報導或本人所稱年齡，與「2026 − 出生年」可能差 ±1）、`approximate_age`（約略值）、`existing_source`（沿用 current 或 11/04 審查）。
- 學生決定檔目前未附來源網址，`source_url` 欄留空待補。
- 另輸出長格式待補清單：每列是一位仍在分析樣本內、但仍缺某個 G1–G5 欄位的候選人，附審查時的查核備註，供分工補查。

```python
from pathlib import Path
import hashlib
import pandas as pd


def find_project_root(start=None):
    start = Path.cwd() if start is None else Path(start)
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate / "AGENTS.md").is_file() and (candidate / "templates").is_dir():
            return candidate
    raise FileNotFoundError("Open this notebook within the student project containing AGENTS.md and templates/.")


PROJECT_ROOT = find_project_root()


def project_file(relative_path, area, must_exist=True):
    # Explicit project-relative paths only; never select the first matching file.
    rel = Path(relative_path)
    if rel.is_absolute() or ".." in rel.parts or "<" in str(rel):
        raise ValueError(f"Replace the placeholder with an exact project-relative path: {relative_path}")
    path = (PROJECT_ROOT / rel).resolve()
    allowed = (PROJECT_ROOT / "data" / area).resolve()
    if allowed == PROJECT_ROOT or not allowed.is_relative_to(PROJECT_ROOT):
        raise ValueError(f"data/{area}/ resolves outside the project root.")
    if not path.is_relative_to(allowed):
        raise ValueError(f"Expected a file inside data/{area}/: {relative_path}")
    if must_exist and not path.is_file():
        raise FileNotFoundError(path)
    return path


def sha256_file(path):
    digest = hashlib.sha256()
    with path.open("rb") as stream:
        for block in iter(lambda: stream.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()

def verify_input_sha256(path, expected):
    # Bind this run to the exact file inspected when the notebook was prepared.
    if (not isinstance(expected, str) or len(expected) != 64
            or any(char not in "0123456789abcdefABCDEF" for char in expected)):
        raise ValueError("Fill the expected SHA-256 from the inspected input before loading data.")
    actual = sha256_file(path)
    if actual != expected.lower():
        raise ValueError(f"Input changed: identify the new version before loading {path.name}.")
    return actual
```

```python
INPUTS = {
    "current": "data/processed/candidates_g1_g6_current.csv",
    "mayor_review": "data/processed/mayor_g1_g6_gap_review_20261104.csv",
    "council_review": "data/processed/council_tk_g1_g5_gap_review_20261104.csv",
    "manual_worklist": "data/processed/manual_check_worklist_taipei_kaohsiung_20261104.csv",
    "student_decisions": "data/raw/student_decisions_20261108/scope_student_decisions_20261108.csv",
}
EXPECTED_INPUT_SHA256 = {
    "current": "790407203ac8d7d4de3bc605fe30fe883007902978f89fa416e7e5593456de02",
    "mayor_review": "6b181cfac8481b00cff19c4207dcf5dc4addc9bfb8c1d17c83f72205d62b225f",
    "council_review": "03773085b6f8ab9fd993ebca376527bbdb130e79898855d9f8a8c8e119ac6769",
    "manual_worklist": "f0aab3e42470046fd1430094b38fb1ce42a34aec69fc9eed6317bc70fb85f377",
    "student_decisions": "1b9d3e63f169aa616a07a1fe3363cc4f3d4a95fb3110e9b0e8f0c61ae7883b23",
}
OUTPUTS = {
    "scope": "data/processed/candidates_scope_g1_g6_20261108.csv",
    "gaps": "data/processed/candidates_scope_g1_g6_gaps_20261108.csv",
    "excluded": "data/processed/candidates_scope_g1_g6_excluded_20261108.csv",
    "g6_worklist": "data/processed/candidates_scope_g6_unknown_worklist_20261109.csv",
}
INPUT_OPTIONS = {key: {"dtype": str, "encoding": "utf-8-sig"} for key in INPUTS}
OUTPUT_READ_OPTIONS = {key: {"dtype": str, "encoding": "utf-8-sig"} for key in OUTPUTS}

input_paths = {}
for key, relative_path in INPUTS.items():
    area = Path(relative_path).parts[1] if len(Path(relative_path).parts) > 1 else ""
    if area not in {"raw", "processed"}:
        raise ValueError(f"Expected a raw or upstream processed input: {relative_path}")
    input_paths[key] = project_file(relative_path, area)
output_paths = {key: project_file(path, "processed", must_exist=False) for key, path in OUTPUTS.items()}
if len(set(output_paths.values())) != len(output_paths):
    raise ValueError("Each output needs a distinct file path.")
if set(input_paths.values()) & set(output_paths.values()):
    raise ValueError("An output must not overwrite one of its inputs.")
existing = [str(path) for path in output_paths.values() if path.exists()]
if existing:
    raise FileExistsError(f"Review existing outputs before replacing them: {existing}")

if set(EXPECTED_INPUT_SHA256) != set(INPUTS):
    raise ValueError("Every input must have its own expected SHA-256.")
input_fingerprints = {key: verify_input_sha256(path, EXPECTED_INPUT_SHA256[key])
                      for key, path in input_paths.items()}
inputs = {key: pd.read_csv(path, **INPUT_OPTIONS.get(key, {})) for key, path in input_paths.items()}
for key, frame in inputs.items():
    print({"input": key, "data_path": INPUTS[key], "rows": len(frame),
           "schema": {column: str(dtype) for column, dtype in frame.dtypes.items()},
           "sha256": input_fingerprints[key]})
```

```python
current = inputs["current"]
assert current["candidate_key"].is_unique

# 1. 分析範圍：全部首長 + 臺北市、高雄市議員
is_head = current["office_type"].isin(["直轄市長", "縣（市）長"])
is_tk_council = (current["office_type"] == "直轄市議員") & current["county_city"].isin(["臺北市", "高雄市"])
scope = current[is_head | is_tk_council].copy().reset_index(drop=True)
print(scope.groupby(["office_type", "county_city"]).size().loc[lambda s: s.index.get_level_values(0) == "直轄市議員"])
print("首長", is_head.sum(), "／範圍合計", len(scope))
assert is_head.sum() == 81 and len(scope) == 280

FIELDS = {"G1": "G1_age_2026_minus_birth_year", "G2": "G2_sex",
          "G3": "G3_education_level", "G5": "G5_first_time_any_elected_office"}
G5_LABEL = {"1": "首次參選任何民選公職", "0": "非首次參選任何民選公職"}
before = scope[list(FIELDS.values())].copy()

# 2. 整理補值來源（只取 accepted 且有建議值者）
review = pd.concat([inputs["mayor_review"].assign(fill_origin="mayor_g1_g6_gap_review_20261104"),
                    inputs["council_review"].assign(fill_origin="council_tk_g1_g5_gap_review_20261104")],
                   ignore_index=True)
print(review.groupby(["field", "decision"]).size())
acc = review[review["decision"].isin(["accepted", "accepted_group_decision"]) & review["proposed_value"].notna()]
acc = acc.assign(value=acc["proposed_value"], status="filled_" + acc["decision"],
                 src=acc["source_url"], note=acc["note"], basis=acc["decision_basis"],
                 evidence=acc["evidence_excerpt"])

wl = inputs["manual_worklist"]
wl_found = wl[wl["found_value"].notna()].copy()
print("人工查核清單已填 found_value：", len(wl_found))
wl_found = wl_found.assign(value=wl_found["found_value"], fill_origin="manual_check_worklist_taipei_kaohsiung_20261104",
                           status="filled_student_check", src=wl_found["source_url"],
                           # 何恕懷：許以琳 2026-11-09 確認粉專與 2026 臺北市第2選區登記者為同一人
                           note=wl_found["checked_by"] + " " + wl_found["checked_date"] + "：" + wl_found["note"].fillna("")
                                + "；許以琳 2026-11-09 確認粉專與登記參選者為同一人",
                           basis="student_manual_check", evidence=wl_found["source_quote"])
fills = pd.concat([acc, wl_found], ignore_index=True)[
    ["candidate_key", "field", "value", "status", "fill_origin", "src", "note", "basis", "evidence"]]
fills = fills[fills["field"].isin(FIELDS)]
assert not fills.duplicated(["candidate_key", "field"]).any(), "同一欄位有兩個補值來源"
unmatched = set(fills["candidate_key"]) - set(scope["candidate_key"])
assert not unmatched, unmatched

# 3. 套用：只填原本空白者
for g, col in FIELDS.items():
    scope[f"{g}_fill_status"] = scope[col].notna().map({True: "current", False: "missing"})
    scope[f"{g}_fill_source"] = pd.Series([None] * len(scope), dtype=object)
    scope[f"{g}_fill_note"] = pd.Series([None] * len(scope), dtype=object)
idx = scope.set_index("candidate_key").index
n_applied = 0
for r in fills.itertuples(index=False):
    i = idx.get_loc(r.candidate_key)
    col = FIELDS[r.field]
    assert pd.isna(scope.at[i, col]), f"{r.candidate_key} {r.field} 原已有值，不覆蓋"
    scope.at[i, col] = r.value
    scope.at[i, f"{r.field}_fill_status"] = r.status
    scope.at[i, f"{r.field}_fill_source"] = r.fill_origin + (f" | {r.src}" if pd.notna(r.src) else "")
    scope.at[i, f"{r.field}_fill_note"] = r.note
    if r.field == "G5":
        scope.at[i, "G5_label"] = G5_LABEL[r.value]
        scope.at[i, "G5_audit_status"] = r.basis
        scope.at[i, "G5_evidence"] = r.evidence
        scope.at[i, "G5_source"] = r.src
    n_applied += 1
print("套用補值", n_applied, "筆")

# 3b. 套用許以琳 2026-11-08 的補值與排除決定
dec = inputs["student_decisions"]
assert set(dec["candidate_key"]) <= set(scope["candidate_key"]), "決定檔有範圍外的 candidate_key"
assert set(dec["action"]) <= {"fill", "exclude", "note"}
# action = note：候選人層級備註（例如原住民、政治家族、爭議事件），寫入 candidate_note
dnote = dec[dec["action"] == "note"].groupby("candidate_key")["note"].agg("；".join)
scope["candidate_note"] = scope["candidate_key"].map(dnote).astype(object)
scope["G1_age_basis"] = scope["G1_fill_status"].map({"missing": float("nan")}).astype(object)
scope.loc[scope["G1_fill_status"] != "missing", "G1_age_basis"] = "existing_source"
dfill = dec[dec["action"] == "fill"]
assert not dfill.duplicated(["candidate_key", "field"]).any()
STUDENT = "student_decisions（許以琳）"
for r in dfill.itertuples(index=False):
    i = idx.get_loc(r.candidate_key)
    note = r.student_source_text + (f"；{r.note}" if pd.notna(r.note) else "")
    if r.field == "G6":
        assert pd.isna(scope.at[i, "G6_political_family"]), f"{r.candidate_name} G6 原已有值"
        scope.at[i, "G6_political_family"] = r.value
        scope.at[i, "G6_label"] = "有三親等內或配偶公職來源記載"
        scope.at[i, "G6_evidence"] = r.student_source_text
        scope.at[i, "G6_source"] = STUDENT
        scope.at[i, "G6_evidence_tier"] = "student_manual_check"
        continue
    col = FIELDS[r.field]
    assert pd.isna(scope.at[i, col]), f"{r.candidate_name} {r.field} 原已有值，不覆蓋"
    scope.at[i, col] = r.value
    scope.at[i, f"{r.field}_fill_status"] = "filled_student_decision"
    scope.at[i, f"{r.field}_fill_source"] = STUDENT
    scope.at[i, f"{r.field}_fill_note"] = note
    if r.field == "G1":
        scope.at[i, "G1_age_basis"] = r.age_basis
        if pd.notna(r.birth_date):
            scope.at[i, "birth_date"] = r.birth_date
            scope.at[i, "birth_year"] = r.birth_date[:4]
            assert 2026 - int(r.birth_date[:4]) == int(r.value)
    if r.field == "G5":
        scope.at[i, "G5_label"] = G5_LABEL[r.value]
        scope.at[i, "G5_evidence"] = r.student_source_text
        if pd.notna(r.source_url):  # 以中選會歷屆候選人檔生日完全吻合核對身分
            scope.at[i, "G5_audit_status"] = "official_historical_record_located_birthdate_match"
            scope.at[i, "G5_source"] = r.source_url
            scope.at[i, f"G5_fill_source"] = STUDENT + " | " + r.source_url
        else:
            scope.at[i, "G5_audit_status"] = "student_decision"
            scope.at[i, "G5_source"] = STUDENT
print("學生補值", len(dfill), "筆")

# 4. 檢查：原有值不變、值域合理
for col in FIELDS.values():
    had = before[col].notna()
    assert (scope.loc[had, col] == before.loc[had, col]).all(), col
assert scope["G2_sex"].dropna().isin(["男", "女"]).all()
assert scope["G3_education_level"].dropna().isin(["研究所", "大學", "大學以下"]).all()
assert scope["G5_first_time_any_elected_office"].dropna().isin(["0", "1"]).all()
ages = pd.to_numeric(scope["G1_age_2026_minus_birth_year"], errors="raise")
assert ages.dropna().between(23, 100).all()

# 4b. 排除：移到 excluded 檔並保留備註，分析檔不含這些人
dexc = dec[dec["action"] == "exclude"]
assert dexc["candidate_key"].is_unique and dexc["note"].notna().all()
excl_mask = scope["candidate_key"].isin(dexc["candidate_key"])
excluded = scope[excl_mask].merge(
    dexc[["candidate_key", "student_source_text", "note", "decided_by", "decided_date"]]
        .rename(columns={"student_source_text": "exclusion_evidence", "note": "exclusion_note"}),
    on="candidate_key", how="left", validate="one_to_one")
n_before = len(scope)
scope = scope[~excl_mask].reset_index(drop=True)
print("排除", len(excluded), "人：", "、".join(excluded["candidate_name"]))
assert len(scope) == n_before - len(dexc)

# 5. 待補清單（長格式）
review_note = review.set_index(["candidate_key", "field"])["note"]
wl_hint = wl.drop_duplicates(["candidate_key", "field"]).set_index(["candidate_key", "field"])["suggested_query"]
rows = []
for r in scope.itertuples(index=False):
    rd = r._asdict()
    for g in FIELDS:
        if rd[f"{g}_fill_status"] == "missing":
            k = (r.candidate_key, g)
            rows.append({"candidate_key": r.candidate_key, "candidate_name": r.candidate_name,
                         "county_city": r.county_city, "office_type": r.office_type,
                         "electoral_district": r.electoral_district, "G4_party": r.G4_party,
                         "field": g, "review_note_20261104": review_note.get(k, float("nan")),
                         "suggested_query": wl_hint.get(k, float("nan"))})
gaps = pd.DataFrame(rows, columns=["candidate_key", "candidate_name", "county_city", "office_type",
                                   "electoral_district", "G4_party", "field", "review_note_20261104", "suggested_query"])
assert len(gaps) == sum((scope[f"{g}_fill_status"] == "missing").sum() for g in FIELDS)

summary = pd.DataFrame({g: scope[f"{g}_fill_status"].value_counts() for g in FIELDS}).fillna(0).astype(int)
print(summary)
print("仍缺：", len(gaps), "個欄位／", gaps["candidate_key"].nunique(), "人")
print(gaps.groupby(["office_type", "county_city", "field"]).size())
print("G6 有記載：", scope["G6_political_family"].notna().sum(), "／未知：", scope["G6_political_family"].isna().sum())

# 6. G6 未知者人工查核清單（回填欄留空）
OFFICE_ORDER = {"直轄市長": 0, "縣（市）長": 1, "直轄市議員": 2}
g6 = scope[scope["G6_political_family"].isna()][
    ["candidate_key", "candidate_name", "office_type", "county_city", "electoral_district", "G4_party", "candidate_note"]].copy()
g6 = g6.sort_values(["office_type", "county_city", "electoral_district", "candidate_name"],
                    key=lambda s: s.map(OFFICE_ORDER) if s.name == "office_type"
                    else s.str.extract(r"第(\d+)選舉區")[0].astype(float) if s.name == "electoral_district" else s).reset_index(drop=True)
for c in ["found_G6_value", "relative_name", "relationship", "relative_office", "source_quote", "checked_by", "checked_date", "note"]:
    g6[c] = float("nan")
assert len(g6) == scope["G6_political_family"].isna().sum() and g6["candidate_key"].is_unique
print("G6 查核清單", len(g6), "人：", g6.groupby("office_type").size().to_dict())

# 與 dtype=str 讀回後一致：缺值統一為 NaN 的字串欄
scope = scope.astype("str")
gaps = gaps.astype("str")
excluded = excluded.astype("str")
g6 = g6.astype("str")
outputs = {"scope": scope, "gaps": gaps, "excluded": excluded, "g6_worklist": g6}
```
