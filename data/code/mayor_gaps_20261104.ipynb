- 仍未知的缺口另列手動查核連結（第 5b 節）。
- 另對 23 人全名比對中選會 1994–2024 歷屆候選人檔，以完整生日或出生年排除同名者。這是補充資訊；「歷屆檔查無本人」**不**當作首次參選證據。

```python
from pathlib import Path
import hashlib, json, csv, re, unicodedata
import pandas as pd
from bs4 import BeautifulSoup


def find_project_root(start=None):
    start = Path.cwd() if start is None else Path(start)
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate / "AGENTS.md").is_file() and (candidate / "templates").is_dir():
            return candidate
    raise FileNotFoundError("Open this notebook within the student project containing AGENTS.md and templates/.")


PROJECT_ROOT = find_project_root()
# SHA-256 after CRLF->LF: identical to the bytes Git stores, so the pins hold on Windows
# checkouts with core.autocrlf=true as well as on LF checkouts.
sha = lambda rel: hashlib.sha256((PROJECT_ROOT / rel).read_bytes().replace(b"\r\n", b"\n")).hexdigest()
norm = lambda s: re.sub(r"\s+", "", unicodedata.normalize("NFKC", s))

INPUTS = {
    "g1_g6_current": "data/processed/candidates_g1_g6_current.csv",
    "mayor_gap_captures": "data/raw/mayor_gaps_20261104/manifest.json",
    "cec_all_offices": "data/raw/cec_all_offices_manifest.json",
    "cec_2022_geo_base": "data/raw/cross_office_round8_20261019/2022_village_elbase.csv",
    "cec_2024_party_list_bulletin": "data/raw/cec_2024_party_list_bulletin.pages.json",
    "udn_wang_zhi_wei": "data/raw/g1_g6_20261103/first_time_8099e6fbb6fb.html",
}
EXPECTED_INPUT_SHA256 = {
    "g1_g6_current": "d8695b306bd7eadbf0de5c8bb2f368c717d23d4ef7ae55cf3cf4b8fa1afbb59c",
    "mayor_gap_captures": "dec5a014e47465223da65b9eee0b145cfd9aab32c6dadcef218e983486eba7c5",
    "cec_all_offices": "9cc85f7f913f5de717ba92ec7698c73cedb29df506fba59a69933d1ac3517231",
    "cec_2022_geo_base": "4d0afbae084ccd77785ebe9a878eadb7e97e828555926d92866dddbfb4b14e79",
    "cec_2024_party_list_bulletin": "ff80083e62ac5d4f4bd82178db224cf985faddfd0d6f1254411679e6a3be42f1",
    "udn_wang_zhi_wei": "6fde37bf170499dc58c01e9816bbbde3d41653e5c7e4865cbd1b11548b5267b2",
}
OUTPUTS = {
    "gap_review": "data/processed/mayor_g1_g6_gap_review_20261104.csv",
    "cec_name_scan": "data/processed/mayor_cec_name_scan_20261104.csv",
    "manual_check_links": "data/processed/mayor_manual_check_links_20261104.csv",
}
for key, rel in INPUTS.items():
    assert sha(rel) == EXPECTED_INPUT_SHA256[key], f"Input changed: {rel}"

# Every capture and every CEC member file is re-hashed against its manifest.
captures = json.loads((PROJECT_ROOT / INPUTS["mayor_gap_captures"]).read_text(encoding="utf-8"))
assert all(c["status"] == "ok" for c in captures)
for c in captures:
    assert sha(c["file"]) == c["sha256_lf"], c["file"]
CAPTURE = {c["label"]: c for c in captures}
cec_manifest = json.loads((PROJECT_ROOT / INPUTS["cec_all_offices"]).read_text(encoding="utf-8"))
# The CEC manifest hashes the original bytes; some originals use CRLF, others LF that
# Git converted to CRLF on checkout. Accept the exact bytes or their LF-normalized form.
for m in cec_manifest["members"]:
    raw = (PROJECT_ROOT / m["path"]).read_bytes()
    assert m["sha256"] in {hashlib.sha256(raw).hexdigest(), sha(m["path"])}, m["path"]
print(len(captures), "captures and", len(cec_manifest["members"]), "CEC manifest entries verified")
for key, rel in OUTPUTS.items():
    assert (PROJECT_ROOT / rel).resolve().is_relative_to((PROJECT_ROOT / "data/processed").resolve())
    assert rel not in INPUTS.values()
```

## 2. 中央社候選人辭典（2026-10-27）

逐人取出生日、學歷、經歷原文，並檢查出生年與主表 G1 是否一致。

```python
def page_text(label):
    soup = BeautifulSoup((PROJECT_ROOT / CAPTURE[label]["file"]).read_bytes(), "html.parser")
    for t in soup(["script", "style", "noscript"]):
        t.decompose()
    return re.sub(r"\s+", " ", soup.get_text(" "))

soup = BeautifulSoup((PROJECT_ROOT / CAPTURE["cna_20261027_mayor_dictionary"]["file"]).read_bytes(), "html.parser")
cna = {}
for box in soup.select("div.dictionary-box"):
    t = re.sub(r"\s+", " ", box.get_text(" ")).strip()
    name = t.split(" ")[0]
    fields = dict(re.findall(r"(生日年月|生日|出生年|黨籍|現職|學歷|經歷)：(.*?)(?= (?:生日年月|生日|出生年|黨籍|現職|學歷|經歷)：| 看更多|$)", t))
    cna[name] = fields
cna_rows = []
for r in targets.itertuples():
    f = cna.get(r.candidate_name, {})
    born = f.get("生日") or f.get("生日年月") or f.get("出生年") or ""
    y = re.match(r"(\d{4})年", born)
    cna_rows.append(dict(candidate_name=r.candidate_name, county_city=r.county_city, G1=r.G1_age_2026_minus_birth_year,
                         cna_birth=born, cna_age=str(2026 - int(y[1])) if y else "", cna_education=f.get("學歷", "")))
cna_view = pd.DataFrame(cna_rows)
mismatch = cna_view[(cna_view.G1 != "") & (cna_view.cna_age != "") & (cna_view.G1 != cna_view.cna_age)]
print("not in CNA dictionary:", cna_view.loc[cna_view.cna_birth.eq("") & cna_view.cna_education.eq(""), "candidate_name"].tolist())
print("G1 vs CNA birth-year mismatches:", len(mismatch))
assert mismatch.empty
cna_view
```

## 3. 中選會 1994–2024 歷屆候選人全名比對

以中央社完整生日（或年月）比對歷屆檔的民國出生年或年月日。只有出生年一致才視為可能同一人；年份不同者為同名他人。

```python
def roc_birth(raw):
    raw = raw.strip().lstrip("'")
    if len(raw) == 3:
        return str(1911 + int(raw)), ""
    if len(raw) == 7:
        return str(1911 + int(raw[:3])), f"{1911 + int(raw[:3])}-{raw[3:5]}-{raw[5:]}"
    return "", ""

name_variants = {n: {n} for n in targets.candidate_name}
name_variants["温世政"].add("溫世政")
lookup = {v: n for n, vs in name_variants.items() for v in vs}
hits = []
for m in cec_manifest["members"]:
    context = m["archive_member"].split("/voteData/", 1)[-1].rsplit("/elcand.csv", 1)[0]
    with (PROJECT_ROOT / m["path"]).open(encoding="utf-8-sig") as fh:
        for line, row in enumerate(csv.reader(fh), 1):
            if len(row) > 12 and row[6].strip() in lookup:
                name = lookup[row[6].strip()]
                year, full = roc_birth(row[9])
                cna_born = cna_view.set_index("candidate_name").at[name, "cna_birth"]
                cna_year = cna_born[:4]
                status = ("birth_year_matches_cna" if year and year == cna_year
                          else "rejected_namesake_birth_year_differs" if year and cna_year else "unresolved_no_birth_to_compare")
                hits.append(dict(candidate_name=name, cec_name=row[6].strip(), election_context=context,
                                 geo_code="|".join(c.strip().lstrip("'") for c in row[:5]), sex_code=row[8].strip().lstrip("'"),
                                 birth_raw=row[9].strip().lstrip("'"), birth_year=year, birth_date=full, education_cec=row[12].strip(),
                                 elected_marker=row[14].strip() if len(row) > 14 else "", cna_birth=cna_born,
                                 identity_status=status, source_file=m["path"], source_row=line))
scan = pd.DataFrame(hits)
scan.insert(0, "candidate_key", scan.candidate_name.map(targets.set_index("candidate_name").candidate_key))
print(scan.identity_status.value_counts().to_dict())
scan[["candidate_name", "election_context", "birth_raw", "cna_birth", "identity_status", "source_file", "source_row"]]
```

在 G5 有缺口的人當中，只有魏平政有出生年一致的命中（蕭燐洪、張靜的命中與主表既有的 G5=0 一致，他們只缺 G3）：2005 年鄉鎮市長選舉，行政區碼 `01|007|01|017|0000`。下一格核對這個代碼是否為彰化縣社頭鄉：2022 年代碼表中，彰化縣社頭鄉為 `10|007|00|170`，舊制三碼 `017` 與新制 `170` 相對應，並以自由時報的報導交叉確認他曾參選鄉長。

```python
matched = scan[scan.identity_status.eq("birth_year_matches_cna")]
g5_gap_names = set(gaps.loc[gaps.field.eq("G5"), "candidate_name"])
# 蕭燐洪 and 張靜 are in scope only for G3; their matches agree with the existing G5 = 0.
assert set(matched.candidate_name) - g5_gap_names == {"蕭燐洪", "張靜"}
assert (targets.set_index("candidate_name").loc[["蕭燐洪", "張靜"], FIELDS["G5"]] == "0").all()
wei = matched[matched.candidate_name.isin(g5_gap_names)]
assert wei.candidate_name.tolist() == ["魏平政"]
w = wei.iloc[0]
assert w.election_context == "2005鄉鎮市長" and w.geo_code == "01|007|01|017|0000" and w.birth_raw == "058" and w.cna_birth == "1969年6月"
geo = list(csv.reader((PROJECT_ROOT / INPUTS["cec_2022_geo_base"]).open(encoding="utf-8-sig")))
assert ["10", "007", "00", "170", "0000", "社頭鄉"] in [[c.strip().lstrip("'") for c in g] for g in geo]
assert "更有人重提他曾選鄉長落敗的往事" in page_text("ltn_5395090_wei_ping_zheng")
print("魏平政：2005 社頭鄉長候選人紀錄與 2026 候選人出生年、鄉鎮代碼、媒體敘述一致")
```

## 4. 逐格判斷與原文核對

每筆證據都指定原始檔與一段原文；下一格會逐筆確認原文確實出現在保存的檔案中，找不到就停止執行。

- **G3 判斷**：只依明載的最高學歷。朱定瑀「工專畢業」、吳淑瑾「大學肄業、高中」、陳盡川「國小畢」、張國威「工業專科學校」都屬「大學以下」；張國威未寫「畢業」，但不論是否畢業都屬大學以下，因此分類不受影響。張火木「研究所畢業」、陳愷璜在北藝大官方頁「畢業於…造型藝術碩士」屬研究所（中央社所列博士班與結業不計）。王志偉的系名加上聯合報「畢業於中國文化大學」屬大學。
- **張靜**：公報與中央社只寫「國立政治大學法律學系」，沒有寫畢業；他曾任法官兼庭長，可推論應已畢業，但這是推論，所以列為 `lead_needs_group_rule`。
- **魏造文 G5**：本人自述 111 年只領表未參選、「這次才算真正參選」，不屬於下方修訂標準的「政治素人／並無政治經歷」，列為 `lead_needs_group_rule`。

### G5 判定標準（2026-11-04 楊喬茵修訂；本縣市首長補查適用）

依序判斷，先符合者為準：

1. **記 0（曾參選）**：中選會歷屆候選人檔有姓名相同、出生年相符的參選紀錄，或本人／媒體明確敘述曾參選民選公職。政務任命（如副縣長）不算民選公職。
2. **記 1（首次參選）**：本人或媒體明確敘述首次參選；**或本人自稱、媒體稱其為「政治素人」；或媒體報導其「並無政治經歷」**。後兩種情形無從另行查核，依報導判定。
3. **記 1（查無歷屆紀錄）**：楊喬茵 2026-11-04 決定：中選會 1994–2024 歷屆候選人檔查無本人（無同名者，或同名者出生年不同）即視為未曾參選。`decision_basis` 記為 `G5_no_prior_cec_record_rule_20261104`。
